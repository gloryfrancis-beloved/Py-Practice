```python
print(f'------------------')
print(f'The Guessing Game')
print(f'------------------')

# goal: guess a secret number, there are three attempts to doing so.
target_number = 4
guesses_left = 3
guessed_correctly = False

while guessed_correctly == False and guesses_left > 0 :
    guess = int(input(f'Guess the secret number, you have only three attempts'))
    if guess == target_number:
        print('Congratulations! You have guessed correctly,YOU WIN!')
        guessed_correctly == True
    else:
        guesses_left = guesses_left - 1
        if guesses_left > 0:
         print(f'You have guessed wrong, you have {guesses_left} attempts left')

if guessed_correctly == False:
   print('You have exhausted your attempts, YOU LOST!')
```

```python
#count vowels

def count_vowels( word):
    count = 0 

    for letter in word:
        if letter in 'aeiou':
            count = count + 1

    return count 

print (count_vowels('beautiful'))
```

```python
# Password Strength Checker 
common_passwords = ['password', '1234567','admin','password123','gory123','000000']

while True:
     is_upper = False
     is_lower = False
     is_digit = False
     password = input('Enter a password (Password must contain at least one capital letter, one small letter and one numerical value),type exit to quit')
     if password.lower() == 'exit':
        break
     elif password in common_passwords:
        print('Password is weak, create a stronger password ')

     for character in password:
        if character.isupper():
            is_upper = True
        if character.islower():
            is_lower = True
        if character.isdigit():
            is_digit = True
     if len(password) < 8:
        print('Password must be at least 8 character long') 
     if not is_upper:
        print('Password must have at least one capital letter')
     if not is_lower:
        print('Password must have at least one small letter')
     if not is_digit:
        print('Password must have at least one digit')

     if len(password) >= 8 and is_upper == True and is_lower == True and is_digit == True:
        print( 'Strong password successfully set')                       
        break
     
        
                 
```

```python
security_logs = {
    '192.168.1.10': {
        'attempts': 2,
        'status': 'safe'
    },
    '192.168.1.20': {
        'attempts': 8,
        'status': 'suspicious',
    },
    '192.168.1.30': {
        'attempts' : 3,
        'status' : 'safe',
    },
    '192.168.1.40': {
        'attempts': 12,
        'status': 'blocked',
    }
}

# Tasks to do

# 1.count the total number of ip addresses
# 2. find the ip with the highest numbers of attempts
# 3. count how many ips are 'safe'
# 4. print a warning for suspicious or blocked ips

total_number = len(security_logs)
print(f'Total IPs : {total_number}')

highest_num = 0
highest_attempts = ''
for key, value in security_logs.items():
    if value['attempts'] > highest_num:
        highest_num = value['attempts']
        highest_attempts = key
print(f'highest attempts :{highest_attempts}')

count = 0
for ip , details in security_logs.items():
    if details['status'] == 'safe':
        count = count + 1
print(f'Safe IPs: {count}')

for ip , details in security_logs.items():
    if details['status'] == 'suspicious':
        print(f'WARNING: {ip} is suspicious')
    elif details['status'] == 'blocked':
        print(f'WARNING: {ip} is blocked')
        
        
        

    

      
      
```
